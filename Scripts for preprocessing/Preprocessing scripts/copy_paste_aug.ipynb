```python
from pathlib import Path
```

```python
labels_path = Path("/home/aayush/Documents/Dataset Manipulation/Edited/A1 Post Merge(Fixed + Removed Classes + Both lanes)/train/labels")
images_path = Path("/home/aayush/Documents/Dataset Manipulation/Edited/A1 Post Merge(Fixed + Removed Classes + Both lanes)/train/images")
```

```python
def bbox_conv(bounding_boxes) -> tuple[float, float, float, float]:
    size = (640, 480)

    cx, cy, w, h = bounding_boxes
    cx *= size[0]
    cy *= size[1]
    w *= size[0]
    h *= size[1]

    xmin = cx - w / 2
    ymin = cy - h / 2
    xmax = cx + w / 2
    ymax = cy + h / 2

    if xmin<0:
        xmin=0

    if ymin<0:
        ymin=0
    
    if xmax>size[0]:
        xmax=size[0]-1
    
    if ymax>size[1]:
        ymax=size[1]-1

    return [xmin, ymin, xmax, ymax]
```

```python
target_id_6 = "6" # Bcz part[0] gives string not int
target_id_5 = "5"
boxes_6 = {}
boxes_5 = {}
for label in labels_path.iterdir():
    with open(label) as f:
        for line in f:
            parts = line.strip().split()
            if parts[0] == target_id_6:
                bbox_6 = list(map(float, parts[1:]))
                if label.stem not in boxes_6:
                    boxes_6[label.stem] = []
                    boxes_6[label.stem].append(bbox_conv(bbox_6))
            elif parts[0] == target_id_5:
                bbox_5 = list(map(float, parts[1:]))
                if label.stem not in boxes_5:
                    boxes_5[label.stem] = []
                    boxes_5[label.stem].append(bbox_conv(bbox_5))
```

```python

from os import mkdir

mkdir("/home/aayush/Documents/Dataset Manipulation/Edited/crops6")
mkdir("/home/aayush/Documents/Dataset Manipulation/Edited/crops5")
```

```python
import cv2

# for crop [ymin:ymax, xmin:xmax]
# how we have [xmin, ymin, xmax, ymax] as a dictonary value

index = 0
for photo in images_path.iterdir():
    if photo.stem in boxes_6:
        for b in range(len(boxes_6[photo.stem])):
            img = cv2.imread(str(photo))
            xmin, ymin, xmax, ymax = map(int, boxes_6[photo.stem][b])
            crop = img[ymin:ymax, xmin:xmax]
            filename = f"/home/aayush/Documents/Dataset Manipulation/Edited/crops6/crop_6_{index}.jpg"
            cv2.imwrite(filename, crop)
            index+= 1

    elif photo.stem in boxes_5:
        for b in range(len(boxes_5[photo.stem])):
            img = cv2.imread(str(photo))
            xmin, ymin, xmax, ymax = map(int, boxes_5[photo.stem][b])
            crop = img[ymin:ymax, xmin:xmax]
            filename = f"/home/aayush/Documents/Dataset Manipulation/Edited/crops5/crop_5_{index}.jpg"
            cv2.imwrite(filename, crop)
            index += 1

```
