```python
import subprocess
from pathlib import Path

def select_img_directory():
    result = subprocess.run(
        ["zenity", "--file-selection", "--directory"],
        stdout=subprocess.PIPE,
        text=True
    )
    return result.stdout.strip()

def select_label_directory():
    result = subprocess.run(
        ["zenity", "--file-selection", "--directory"],
        stdout=subprocess.PIPE,
        text=True
    )
    return result.stdout.strip()

path_img = Path(select_img_directory())
print(str(path_img))

path_label = Path(select_label_directory())
print(str(path_label))
```

```python
filenames = []

with open("Upload_log.txt", "r") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue

        # remove "Uploading " and "..."
        filename = line.strip().split()[-1]
        filenames.append(filename)

print(filenames)
print(len(filenames))
```

```python
from pathlib import Path

i = 0
filenames = set(filenames)  # Convert list to set for faster lookup
for img_path in path_img.glob("*.jpg"):
    label_path = path_label / (img_path.stem + ".txt")

    if not label_path.exists():
        print(f"Missing label for {img_path.name}, skipping")
        continue

    elif img_path.name in filenames:
        print(f"{img_path.name} already uploaded, skipping")
        continue
    
    print(f"{i}. Uploading {img_path.name}")

    project.upload(
        image_path=str(img_path),
        annotation_path=str(label_path),
        split="train",
        num_retry_uploads=3
    )

    i += 1
```
